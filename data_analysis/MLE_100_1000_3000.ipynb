```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

df = pd.read_csv("Artworks_dates_analysis.csv", low_memory=False)
df["year_mid"] = pd.to_numeric(df["year_mid"], errors="coerce")
df = df.dropna(subset=["year_mid"])

sample = df.sample(n=100, replace=True, random_state=100)
years = sample["year_mid"]

mu = years.mean()
sigma = 34.05

print("Sample size:", len(years))
print("MLE mu:", round(mu, 2))
print("Fixed sigma:", sigma)

x = np.linspace(years.min() - 20, years.max() + 20, 400)
y = norm.pdf(x, loc=mu, scale=sigma)

plt.hist(years, bins=20, density=True, color="lightblue", edgecolor="white")
plt.plot(x, y, color="orangered", linewidth=2, label="Fitted Normal")
plt.xlabel("Creation year")
plt.ylabel("Density")
plt.title("Normal MLE: n = 100, mu = " + str(round(mu, 2)))
plt.legend()
plt.tight_layout()
plt.savefig("mle_100.png", dpi=150)
plt.show()
```

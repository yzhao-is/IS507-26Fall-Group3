```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Artworks_dates_analysis.csv", low_memory=False)
df["year_mid"] = pd.to_numeric(df["year_mid"], errors="coerce")
df = df.dropna(subset=["year_mid"])

full_mean = df["year_mid"].mean()
means = []

for i in range(500):
    sample = df.sample(n=100, replace=True, random_state=100 + i)
    means.append(sample["year_mid"].mean())

    if i == 0:
        sample.to_csv("sample_100.csv", index=False)

results = pd.DataFrame({"draw": range(1, 501), "sample_mean": means})
results["absolute_error"] = abs(results["sample_mean"] - full_mean)
results.to_csv("sampling_100_500_means.csv", index=False)

print("Sample size:", 100)
print("Number of draws:", 500)
print("Full-data mean:", round(full_mean, 6))
print("Sampling SD:", round(results["sample_mean"].std(), 2))
print("Mean absolute error:", round(results["absolute_error"].mean(), 2))

plt.hist(means, bins=20, color="tomato", edgecolor="white")
plt.axvline(full_mean, color="darkblue", linestyle="--", label="Full-data mean")
plt.xlabel("Mean creation year")
plt.ylabel("Number of samples")
plt.title("Sampling: n = 100 (500 draws)")
plt.legend()
plt.tight_layout()
plt.savefig("sampling_100.png", dpi=150)
plt.show()
```

```python
import pandas as pd
```

```python
df = pd.read_csv("train.csv")
```

```python
# Lable unassigned and missing value as “Unknown”
clean = df["Classification"].astype("string").str.strip()

df["Classification_clean"] = (
    clean.replace({"(not assigned)": pd.NA, "": pd.NA})
         .fillna("Unknown")
)

print(df["Classification_clean"].value_counts())
print("Total records:", len(df))
print("Unknown records:", df["Classification_clean"].eq("Unknown").sum())

df.to_csv("Classification_train_cleaned.csv", index=False, encoding="utf-8-sig")
```
